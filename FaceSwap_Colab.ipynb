```python
# 2) Ativar o link público da interface (patch share = True)
import pathlib
target = pathlib.Path('/content/facefusion/facefusion/uis/layouts/default.py')
txt = target.read_text()
old = "ui.launch(favicon_path = 'facefusion.ico', inbrowser = state_manager.get_item('open_browser'))"
new = "ui.launch(favicon_path = 'facefusion.ico', inbrowser = state_manager.get_item('open_browser'), share = True)"
if new not in txt and old in txt:
    target.write_text(txt.replace(old, new))
    print('Patch aplicado: link publico ativado')
elif new in txt:
    print('Patch ja estava aplicado')
else:
    print('ATENCAO: padrao nao encontrado, verifique a versao do FaceFusion')
```
